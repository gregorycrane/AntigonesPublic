```python
import re
print(len(card_alignment))

basecards = {}
curchap = ''
curcard = ''
f = open('/Users/gcrane/github/canonical-greekLit/data/tlg0086/tlg034/tlg0086.tlg034.perseus-grc2.xml')
for l in f:
    m = re.search('<milestone unit="card" n="([^"]+)',l)
    if(m):
        basecards[m[1]] = 1

    m = re.search('type="chapter"[^>]+n="([^"]+)',l)
    if(m):
        curchap = m[1]
    m = re.search('type="subchapter"[^>]+n="([^"]+)',l)
    if(m):
        curcard = curchap + '.' + m[1]
        basecards[curcard] = 1

newcards = {}

def additem(curdict,curitem):
    if(curitem in curdict):
        curdict[curitem] = curdict[curitem] + 1
    else:
        curdict[curitem] = 1
        

for foo in card_alignment:
    additem(newcards,str(card_alignment[foo]))

for foo in newcards:
    if(newcards[foo] > 1):
        print('extra1',foo,newcards[foo])
    if(not str(foo) in basecards):
        print('extra2',foo)

i = 0
for foo in basecards:
    if(not foo in newcards):
        print('missing',foo)
    else:
        i = i + 1
        #print(i,'confirmed',foo)
    
```
