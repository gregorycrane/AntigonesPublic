```python
import re

basename = 'tlg0012.tlg002.popewatson1858-eng2.xml'
fname = '/Users/gcrane/github/Homerica/data/tlg0012/tlg002/' + basename
outfname = '/Users/gcrane/github/AntigonesPublic/staging/' + re.sub('\.xml','-2.xml',basename)

f = open(fname)
for l in f:
    if(re.search('<note ',l) and not re.search('</note>',l)):
        print(l)
f.close()

f = open(fname)
workbuf = f.read()
f.close()
notelist = {}
reflist = {}
while(re.search('(<note xml:id="([^>]+)">.+</note>)',workbuf)):
    m = re.search('(<note xml:id="([^>]+)">.+</note>)',workbuf)
    #print(m[2],m[1])
    notelist[m[2]] = m[1]
    workbuf = re.sub('(<note xml:id="([^>]+)">.+</note>)','',workbuf,1)

while(re.search('(<ref target="(note[^"]+)"/>)',workbuf)):
    m = re.search('(<ref target="(note[^"]+)"/>)',workbuf)
    #print(m[2],m[1])
    reflist[m[2]] = m[1]
    curkey = m[2]
    workbuf = re.sub('(<ref target="(note[^"]+)"/>)',notelist[curkey],workbuf,1)

print(len(reflist),len(notelist))

for foo in notelist:
    if(not foo in reflist):
        print('refmiss',foo)

outf = open(outfname,'w')
outf.write(workbuf)
outf.close()

print(fname,outfname)
```
